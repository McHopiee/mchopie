# Building for the Poway Symphony Orchestra: A Journey in Tech, Music, and Purpose

When we were presented with several nonprofit organizations to build for, the Poway Symphony Orchestra stood out immediately. Music is a genuine part of who we are, and using our skills in technology to support an organization that shares that same passion felt like the perfect fit.

Choosing an organization we truly cared about made all the difference. It pushed us to work harder and think more carefully about every decision we made throughout the project.

---

### Building Something We Were Proud Of

We had big ideas from the start - member cards, a sign-in page, a membership registration flow, an admin page, an admin chat, games, and musician highlights. We wanted the website to feel engaging and welcoming for the community.

Throughout the process, we checked in regularly with Mr. Mortensen, and we were thrilled to see that he was genuinely impressed with our progress. His feedback kept us motivated and on track.

Even with a strong prototype built, the vision is far from finished. We still have plenty of ideas for what we want to add, and we're excited to keep building on what we've started.

---

### Presenting to the Poway Symphony Orchestra

We presented our project to Ms. Schafer and Mr. Reid of the Poway Symphony Orchestra, and it went incredibly well. Sharing our work with the very people we built it for was a rewarding experience we won't forget.

We also had the opportunity to present to professionals in the technology industry. The support and encouragement we received from both audiences was truly motivating and pushed us to think even bigger.

---

### After the Curtain Call

As a token of their gratitude, Ms. Schafer and Mr. Reid offered us tickets to a Poway Symphony Orchestra concert - and we had a wonderful time. Hearing the music live brought the whole journey full circle.

We are so grateful for everyone who supported us along the way - from Mr. Mortensen to Ms. Schafer and Mr. Reid to the technology professionals who gave us their time. Learning that they enjoyed our project made every bit of effort worth it, and we leave this experience feeling inspired to keep growing and pursuing what we love.

### Designing The Website

<a href="https://pso.opencodingsociety.com/" target="_blank" rel="noopener noreferrer" style="display:inline-block; padding:14px 24px; font-size:1.1rem; font-weight:700; background:#0b5ed7; color:#ffffff; border-radius:10px; text-decoration:none; margin:8px 0;">Visit PSO Website</a>

The website was designed to match Open Coding Society branding while creating a modern, welcoming experience for visitors. The homepage uses a dynamic video background and a clear visual hierarchy so users can quickly understand what the organization offers and where to go next.

- Color palette and visual styling were aligned with existing OCS branding for cross-platform consistency.
- A dynamic hero/video element was used to create a stronger first impression than a static layout.
- Core areas like learning pathways, projects, and opportunities were prioritized in layout order.
- Spacing, section structure, and calls to action were organized to keep navigation intuitive.

### Designing And Creating The Backend

<a href="https://wifighters.opencodingsociety.com/" target="_blank" rel="noopener noreferrer" style="display:inline-block; padding:14px 24px; font-size:1.1rem; font-weight:700; background:#0b5ed7; color:#ffffff; border-radius:10px; text-decoration:none; margin:8px 0;">Visit PSO Backend</a>

The backend admin experience was built around a dashboard-first model so administrators can immediately see critical platform activity after login. Management tools are centralized and grouped by workflow to make day-to-day operations faster, clearer, and easier to scale.

- Top-level metrics include pending requests, approvals, users, member cards, leaderboard data, and chat activity.
- Related admin tasks are grouped together: membership requests, user accounts, roster management, profile cards, and communication.
- Information architecture was designed for quick decisions and lower overhead during routine tasks.
- Emphasis on usability reduces onboarding time for new admins and minimizes errors in high-volume workflows.

### Member Cards

<a href="https://pso.opencodingsociety.com/musicians/" target="_blank" rel="noopener noreferrer" style="display:inline-block; padding:14px 24px; font-size:1.1rem; font-weight:700; background:#0b5ed7; color:#ffffff; border-radius:10px; text-decoration:none; margin:8px 0;">View Member Cards</a>

![Work On Website Card]({{ site.baseurl }}/images/personal/card.png)

The Member Cards section highlights individual musicians while helping visitors better understand the orchestra as a whole. It combines profile customization, instrument discovery, and searchable roster tools to improve both member visibility and community engagement.

- Visitors can explore instrument groups and connect instrument names with sample sounds.
- Search and filtering make it easier to find members as the roster grows.
- Personalized cards let members showcase involvement beyond basic roster information.
- Planned enhancements include recordings, profile pictures, and richer biographies.
- Individual profiles strengthen recognition, identity, and interaction across the orchestra community.

### _**Photo Caption**_

I am near the front row and I took this photo before the presentation started so as not to be disruptive and take pictures during the presentation. This proves my experience because I took a photo of all the people attending as well as Mr. Mort and Mr. Johnson.

### _**Before the Event**_
**What did you expect before attending?**

I expected that there were not going to be a lot of people attending the event, but the whole classroom was filled with people. I expected it to be professional and inspiring, and that is exactly what it was.

**What were you curious about?**

I was curious about what exactly he did as a CEO, what inspired him to go through with his ideas, and what he did to make it successful.

### _**What I Saw**_
Slides with pictures and minimal text, mostly Mr. Johnson telling a story about his company and how he became CEO, and what his company does.

The most interesting thing I heard was that in the competition they competed in, their product actually lost to a massage chair. Another interesting thing I heard was that they did an event in front of elementary schoolers where they taught them how to vote through pizza.

### _**What I Learned**_
Something specific I learned is that one of the best ways to pitch a product to somebody is to start with something easy. Have the person you are pitching to know that they can talk to you, then you can hold a conversation. Start with the easy things, then go onto the more in-depth details.

### _**What Surprised Me**_
Something that surprised me was that Mr. Johnson brought his wife and child. It was nice to see them.

Something else that surprised me was how passionate he was about what he does and how encouraging he is. He explained that if someone has a dream, they should keep believing in it until they get to where they want to be.

### _**What Inspired Me**_
What inspired me was his talk about dreams and goals. Mr. Johnson explained that when you are pitching your idea or product to people and some say no, that does not mean that you stop pursuing what you want to do. You keep going and improving until you get someone who believes in you, someone who will say "yes" to your product or pitch.

**Why did this connect to your personal interests?**

This is connected to my personal interests because I want to go into computer science in college and in the future. I know that it is currently a very hard major to get into, specifically because so many other people want to do it and because AI is hard to compete with. But knowing that I know how to use AI, and that I will eventually learn more skills to get myself out there, inspired me to continue pursuing computer science.

### _**Interaction**_
**Who did you talk to or learn from?**

Speaker: Elijah Johnson, entrepreneur, former Division I football player, and University of San Diego graduate. He is the Founder and CEO of SafeBallot, a secure voting technology platform focused on improving accessibility, transparency, and trust in elections.

**What did they explain or show you?**

He explained multiple ways to successfully pitch ideas and present to others. He described his journey to becoming the CEO and founder of a company, and how he started from a small shoe-cleaning business to building a successful company with great ideas.

### _**Depth of Experience**_
**How did this event change your thinking?**

This event changed my thinking by changing my view of myself and my goals. Before, I was very unsure of what I wanted to do in the future because I was comparing myself to other people. Would I get into this college as a computer science major? It changed this thinking, turning this doubt into a reason to do something about it. I should get myself out there and find something to do so I can be a better candidate in the future.

**How does this connect to your CS class, project, or future goals?**

This connects to my CS future goals because, as I mentioned, I want to do something in computer science in the future. Specifically, I want to do something in game development. Going to this presentation really helped me look toward the future and motivated me to keep working toward my goals.

### _**Final Reflection**_
**Overall, why was this event meaningful to you?**

This event was meaningful to me because it told me why I should start before I am ready.
