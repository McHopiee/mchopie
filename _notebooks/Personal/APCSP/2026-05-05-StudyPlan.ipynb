### Incorrect Questions by Topic

#### 1. Internet Concepts (4.1)
**Q4 - Relationship between Internet and World Wide Web**  
![Q4 missed question]({{ site.baseurl }}/images/personal/image.png)

**What this question tested:** precise vocabulary.  
The correct choice is that the Internet is a *network of interconnected networks*, while the Web is a *system of linked resources* accessed through the Internet.

**How my answer was incorrect:** choice B sounds technical (servers, devices), but it incorrectly defines the Web as a device network.

**Takeaway:** when you see Internet vs. Web, map it as *infrastructure vs. information system*.

---

#### 2. Fault Tolerance / Networking (4.2)
**Q17 - Disconnecting devices B and C**  
![Q17 missed question]({{ site.baseurl }}/images/personal/image2.png)

**What this question tested:** minimum links to remove to break all communication paths.

**Why three felt correct:** removing three obvious links can break direct-looking routes, but at least one alternate path still remains.

**Why four is correct:** B and C stay connected through redundancy unless one more edge is removed. A reliable strategy is to isolate one endpoint by counting/removing all links needed to prevent any route from leaving that node and reaching the other.

**Takeaway:** in network cut questions, do not stop after breaking major routes; verify that *no* backup path survives.

---

#### 3. Public Key Encryption (5.6)
**Q32 - Statement about public key encryption**  
![Q32 missed question]({{ site.baseurl }}/images/personal/image3.png)

**What this question tested:** asymmetric-key roles and why public-key crypto is secure.

**How my answer was incorrect:** choice B assumes that if a public key is intercepted, security fails.

**Why that is incorrect:** public keys are meant to be shared. Security depends on keeping the private key secret. Public-key encryption is specifically useful over open channels with eavesdroppers.

**Takeaway:** remember the rule: *public key encrypts, private key decrypts*.

---

#### 4. Iteration + Lists (3.3 / 3.10)
**Q42 - Iterate over integerList**  
![Q42 missed question]({{ site.baseurl }}/images/personal/image4.png)

**What this question tested:** tracing accumulation with `item MOD 2`.

For list `[4, 2, 5, 4, 2, 3, 1]`, the remainders are `0, 0, 1, 0, 0, 1, 1`, which sum to `3`.

**Why my answer was incorrect:** choosing 4 reflects counting even numbers instead of summing odd-indicator remainders.

**Takeaway:** in loop-trace questions, write the running value each iteration so operation meaning (`MOD 2`) stays explicit.

---

#### 5. Boolean + Conditionals (3.5 / 3.6)
**Q65 - Hot and humid conditionals**  
![Q65 missed question]({{ site.baseurl }}/images/personal/image5.png)

**What this question tested:** multi-select Boolean evaluation with `hot = true`, `humid = false`.

Correct choices are **B and C**:
- B executes because `NOT humid` is true, then displays `hot OR humid`, which is true.
- C executes because `hot OR humid` is true, then displays `hot`, which is true.

**I quite literally just failed to choose a second option**
