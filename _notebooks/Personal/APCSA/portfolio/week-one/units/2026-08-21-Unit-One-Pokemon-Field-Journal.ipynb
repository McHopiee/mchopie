## Part 3 - Professor Oak's Potion Lab
**Topics:** 1.5, 1.6, 1.11

Use arithmetic and bounds logic to cap HP after healing.

```text
Bulbasaur HP: 34
Maximum HP: 45
Potion restores 20 HP
New HP: 45
```

{% capture unit1_potion_lab %}
int hp = 34;
int maxHP = 45;
hp += 20;

if (hp > maxHP) {
  hp = maxHP;
}

System.out.println("New HP: " + hp);
{% endcapture %}

## Part 7 - The Pokemon Math Lab
**Topic:** 1.11

Browser runner adaptation (JavaScript) to demonstrate math helpers directly.

const attack = 72;
const defense = 51;
const power = 60;

const rawDamage = attack + power - defense;
const boundedDamage = Math.max(1, Math.min(rawDamage, 120));
const critMultiplier = Math.sqrt(1.21);
const encounter = Math.random();

console.log("Calculated damage:", Math.floor(boundedDamage * critMultiplier));
console.log("Encounter roll:", encounter.toFixed(4));

## Final Project - Build Your Own Pokemon

Professor Oak has asked you to create a new Pokemon for the Pokedex.

### Requirements
Your Pokemon must have:
- Name
- Type
- Level
- HP
- Attack
- Defense
- At least 2 moves

Your program must:
1. Create the Pokemon object.
2. Display its Pokedex entry.
3. Allow the Pokemon to use a move.
4. Calculate something using logic.
5. Modify one of its stats.
6. Use at least one void method.
7. Use at least one non-void method.
8. Use multiple String operations.
9. Include comments explaining important sections.
10. Include user input for the trainer's name.

{% capture unit1_final_demo %}
Scanner scanner = new Scanner(System.in);

System.out.println("Enter trainer name:");
String trainer = scanner.nextLine();
String pokemon = "Floraclaw";
int hp = 67;
int attack = 74;
int defense = 59;

System.out.println("Trainer: " + trainer);
System.out.println("Pokemon: " + pokemon);
System.out.println("HP: " + hp);
System.out.println("Attack: " + attack);
System.out.println("Defense: " + defense);

int damage = attack - defense + 8;
if (damage < 1) {
  damage = 1;
}
System.out.println(pokemon + " used Vine Lash!");
System.out.println("Damage: " + damage);

hp += 12;

  hp = 100;
}
