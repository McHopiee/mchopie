## Unit 1 - Primitive Types

### 1. Starter Registration
**Topics:** variables, literals, primitive data types, expressions, and output

Professor Oak records a new trainer. Every variable has a declared type: `int` stores whole numbers, `double` stores decimals, and `boolean` stores `true` or `false`.

```text
PROFESSOR OAK'S LAB

Trainer: Hope
Age: 16
Starter: Charmander
Ready for training: true
```

```java
String trainer = "Hope";
int age = 16;
String starter = "Charmander";
boolean ready = true;

System.out.println("Trainer: " + trainer);
System.out.println("Age: " + age);
System.out.println("Starter: " + starter);
System.out.println("Ready for training: " + ready);
```

**Trace check:** Why is `age` an `int`, while `ready` is a `boolean`? What value does `7 / 2` produce in Java?

## Unit 2 - Using Objects

### 4. Meet the Pokemon
**Topics:** classes, objects, constructors, references, and instance methods

A class is a blueprint. An object is one Pokemon created from that blueprint. The variable stores a reference to the object.

```java
public class Pokemon {
    private String name;
    private String type;
    private int level;

    public Pokemon(String name, String type, int level) {
        this.name = name;
        this.type = type;
        this.level = level;
    }

    public String getName() {
        return name;
    }
}

Pokemon starter = new Pokemon("Charmander", "Fire", 5);
System.out.println(starter.getName());
```

`new` calls the constructor and creates the object. `starter` is a reference variable that points to that object.

### 5. Pokedex Documentation
**Topics:** method calls, parameters, return values, and comments

Methods let an object perform an action or answer a question. A `void` method performs an action; a non-void method returns a value.

```java
// A void method call performs an action.
System.out.println("Charmander used Ember!");

// A non-void method returns a value that can be stored.
String entry = "Charmander";
int nameLength = entry.length();
System.out.println("Name length: " + nameLength);
```

Comments explain intent to future trainers and teammates. They do not change what the program does.

## Unit 3 - Boolean Expressions and Selection

### 7. Battle Math
**Topics:** relational operators, boolean operators, and precedence

A boolean expression always evaluates to `true` or `false`.

{% capture unit3_boolean_demo %}
int hp = 28;
int maxHP = 45;
boolean lowHP = hp < 15;
boolean canBattle = hp > 0 && !lowHP;
boolean needsHelp = hp <= 0 || hp < maxHP / 2;

System.out.println("Low HP: " + lowHP);
System.out.println("Can battle: " + canBattle);
System.out.println("Needs help: " + needsHelp);
{% endcapture %}
{% include runners/code-java.html runner_id="unit3-boolean-demo" code=unit3_boolean_demo %}

Java evaluates `!` before `&&`, and `&&` before `||`. Short-circuiting means `A && B` stops when `A` is false, while `A || B` stops when `A` is true.

## Unit 4 - Iteration

### 10. Training Loops
**Topics:** `while` loops, `for` loops, nested loops, tracing, and off-by-one errors

Loops repeat instructions while a condition remains true. A `while` loop is useful when the number of repetitions is not known in advance; a `for` loop is useful when the bounds are clear.

{% capture unit4_loops_demo %}
int experience = 0;
int battles = 0;

while (battles < 3) {
  experience += 20;
  battles++;
}

System.out.println("Experience after training: " + experience);

for (int level = 1; level <= 3; level++) {
  System.out.println("Training level " + level);
}
{% endcapture %}
{% include runners/code-java.html runner_id="unit4-loops-demo" code=unit4_loops_demo %}

### Loop tracing challenge

For each outer-loop iteration, the inner loop completes all of its iterations.

```java
for (int row = 1; row <= 2; row++) {
    for (int badge = 1; badge <= 3; badge++) {
        System.out.println("Row " + row + ", badge " + badge);
    }
}
```

**Trace checks:** The nested loop prints 6 lines. Changing `badge <= 3` to `badge < 3` would print 4 lines. A loop variable must change toward the stopping condition, or the loop can run forever. String indexes likewise stop at `i < name.length()`, never `i <= name.length()`.

## Final Project - Build Your Own Pokemon

Your program must:
- Declare a name, type, level, HP, attack, defense, and at least two moves.
