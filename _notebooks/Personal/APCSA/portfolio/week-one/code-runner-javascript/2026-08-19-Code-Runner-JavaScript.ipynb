### Runner 3 — Squirtle's Adventure

Make choices and follow Squirtle's branching lake adventure.

{% capture squirtle_adventure %}
console.log("💧 Squirtle's Adventure");
console.log("Squirtle is standing at the edge of a lake.");

const lakeChoice = prompt(
  "What should Squirtle do?\n1. Swim across the lake\n2. Explore the forest\n3. Rest by the water"
);

if (lakeChoice === "1") {
  console.log("Squirtle jumped into the water!");
  console.log("You see something shiny at the bottom...");

  const diveChoice = prompt("Do you:\n1. Dive down\n2. Keep swimming");
  if (diveChoice === "1") {
    console.log("Squirtle found a Water Stone!");
  } else {
    console.log("Squirtle kept swimming and found a quiet cove.");
  }
} else if (lakeChoice === "2") {
  console.log("Squirtle explored the forest and met a friendly Bulbasaur!");
} else if (lakeChoice === "3") {
  console.log("Squirtle rested by the water and watched the waves sparkle.");
} else {
  console.log("Squirtle waited for a trainer to choose an adventure.");
}
{% endcapture %}
