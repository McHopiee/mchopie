### Runner 2 — Choose Your Attack

Now they're **choosing what happens**. This introduces `Scanner` for input and `if`/`else if`/`else` for branching.

{% capture java_choose_attack %}
Scanner scanner = new Scanner(System.in);

System.out.println("Charmander is ready for battle!");
System.out.println("Choose an attack:");
System.out.println("1. Scratch");
System.out.println("2. Ember");
System.out.println("3. Growl");

int choice = scanner.nextInt();

if (choice == 1) {
  System.out.println("Charmander used Scratch!");
} else if (choice == 2) {
  System.out.println("Charmander used Ember!");
} else if (choice == 3) {
  System.out.println("Charmander used Growl!");
} else {
  System.out.println("Charmander doesn't know that move!");

### Runner 3 — Charmander Battle

Taking that concept further with a `while` loop: Charmander keeps battling Pidgey, round after round, until one of them faints.

{% capture java_battle %}
Scanner scanner = new Scanner(System.in);

int charmanderHP = 50;
int pidgeyHP = 40;

System.out.println("A wild Pidgey appeared!");

while (charmanderHP > 0 && pidgeyHP > 0) {
  System.out.println();
  System.out.println("Charmander HP: " + charmanderHP);
  System.out.println("Pidgey HP: " + pidgeyHP);
  System.out.println("Choose your move:");
  System.out.println("1. Scratch");
  System.out.println("2. Ember");
  System.out.println("3. Growl");

  int choice = scanner.nextInt();
  int damage = 0;

  if (choice == 1) {
    System.out.println("Charmander used Scratch!");
    damage = 8;
  } else if (choice == 2) {
    System.out.println("Charmander used Ember!");
    System.out.println("It's super effective!");
    damage = 18;
  } else if (choice == 3) {
    System.out.println("Charmander used Growl!");
    damage = 0;
  } else {
    System.out.println("Charmander doesn't know that move!");
  }

  pidgeyHP = pidgeyHP - damage;

  if (damage > 0) {
    System.out.println("Pidgey lost " + damage + " HP!");
  }

  if (pidgeyHP <= 0) {
    System.out.println("Pidgey fainted!");
    System.out.println("You won!");
    break;
  }
}
{% endcapture %}
