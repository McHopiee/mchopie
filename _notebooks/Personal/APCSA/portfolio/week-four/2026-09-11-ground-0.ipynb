(function () {
  var tabs = Array.prototype.slice.call(document.querySelectorAll('.ground-zero-tab'));
  var panels = Array.prototype.slice.call(document.querySelectorAll('.ground-zero-panel'));
  function activateTab(tab) {
    tabs.forEach(function (item) {
      var selected = item === tab;
      item.setAttribute('aria-selected', selected ? 'true' : 'false');
      item.tabIndex = selected ? 0 : -1;
    });
    panels.forEach(function (panel) {
      panel.hidden = panel.id !== tab.getAttribute('aria-controls');
    });
    tab.focus();
  }
  tabs.forEach(function (tab, index) {
    tab.addEventListener('click', function () { activateTab(tab); });
    tab.addEventListener('keydown', function (event) {
      var nextIndex = index;
      if (event.key === 'ArrowRight') nextIndex = (index + 1) % tabs.length;
      if (event.key === 'ArrowLeft') nextIndex = (index - 1 + tabs.length) % tabs.length;
      if (nextIndex !== index) {
        event.preventDefault();
        activateTab(tabs[nextIndex]);
      }
    });
  });
})();
</script>

<script>
(function () {
  var storageKey = 'mchopie-ground-zero-technical-evaluation-v1';
  var savedValues = {};
  try {
    savedValues = JSON.parse(localStorage.getItem(storageKey) || '{}');
  } catch (error) {
    savedValues = {};
  }

  var evaluationPanel = document.querySelector('.ground-zero-evaluation');
  if (evaluationPanel) {
    var evaluationHeading = evaluationPanel.querySelector('h2');
    if (evaluationHeading && !evaluationPanel.querySelector('.ground-zero-scale-note')) {
      var scaleNote = document.createElement('p');
      scaleNote.className = 'ground-zero-scale-note';
      scaleNote.innerHTML = '<strong>Self Rank, Peer, and Teacher all use the same 0–5 scale.</strong><br>0 = Not Entered · 1 = Explore · 2 = Understand · 3 = Apply · 4 = Analyze / Evaluate · 5 = Create / Extend';
      evaluationHeading.insertAdjacentElement('afterend', scaleNote);
    }
  }

  document.querySelectorAll('.ground-zero-evaluation table').forEach(function (table) {
    var headers = Array.prototype.map.call(table.querySelectorAll('thead th'), function (header) {
      return header.textContent.trim();
    });
    if (headers.indexOf('Self Rank') === -1) return;

    table.querySelectorAll('tbody tr').forEach(function (row) {
      var cells = row.querySelectorAll('td');
      var rubricId = cells[0] && cells[0].textContent.trim();
      if (!rubricId) return;

      ['Self Rank', 'Peer', 'Teacher'].forEach(function (field) {
        var column = headers.indexOf(field);
        var cell = cells[column];
        if (!cell) return;
        var key = rubricId + '-' + field.toLowerCase().replace(' ', '-');
        var input = document.createElement('input');
        input.className = 'rubric-input';
        input.type = 'number';
        input.min = '0';
        input.max = '5';
        input.step = '1';
        input.inputMode = 'numeric';
        input.value = Object.prototype.hasOwnProperty.call(savedValues, key) ? savedValues[key] : (cell.textContent.trim() === '—' ? '' : cell.textContent.trim());
        input.setAttribute('aria-label', rubricId + ' ' + field + ' ranking, 0 through 5');
        input.placeholder = '0–5';
        input.addEventListener('input', function () {
          var numericValue = Number(input.value);
          if (input.value !== '' && (!Number.isInteger(numericValue) || numericValue < 0 || numericValue > 5)) return;
          savedValues[key] = input.value;
          localStorage.setItem(storageKey, JSON.stringify(savedValues));
        });
        cell.textContent = '';
        cell.appendChild(input);
      });
    });
  });
})();
</script>

  var storageKey = 'mchopie-ground-zero-professional-evaluation-v1';
  var savedValues = {};
  try { savedValues = JSON.parse(localStorage.getItem(storageKey) || '{}'); } catch (error) { savedValues = {}; }
  professionalPanel.querySelectorAll('table').forEach(function (table) {
    var headers = Array.prototype.map.call(table.querySelectorAll('thead th'), function (header) { return header.textContent.trim(); });
    if (headers.indexOf('Self Rank') === -1) return;
    table.querySelectorAll('tbody tr').forEach(function (row) {
      var cells = row.querySelectorAll('td');
      var rubricId = cells[0].textContent.trim();
      ['Self Rank', 'Peer', 'Teacher'].forEach(function (field) {
        var column = headers.indexOf(field);
        var cell = cells[column];
        var key = rubricId + '-' + field.toLowerCase().replace(' ', '-');
        var input = document.createElement('input');
        input.className = 'rubric-input';
        input.type = 'number';
        input.min = '0'; input.max = '5'; input.step = '1'; input.inputMode = 'numeric';
        input.value = Object.prototype.hasOwnProperty.call(savedValues, key) ? savedValues[key] : (cell.textContent.trim() === '—' ? '' : cell.textContent.trim());
        input.placeholder = '0–5';
        input.setAttribute('aria-label', rubricId + ' ' + field + ' ranking, 0 through 5');
        input.addEventListener('input', function () {
          if (input.value !== '' && (!Number.isInteger(Number(input.value)) || Number(input.value) < 0 || Number(input.value) > 5)) return;
          savedValues[key] = input.value;
          localStorage.setItem(storageKey, JSON.stringify(savedValues));
        });
        cell.textContent = ''; cell.appendChild(input);
      });
    });
  });

  var tabs = Array.prototype.slice.call(document.querySelectorAll('.ground-zero-tab'));
  var panels = Array.prototype.slice.call(document.querySelectorAll('.ground-zero-panel'));
  function activateTab(tab) {
    tabs.forEach(function (item) { var selected = item === tab; item.setAttribute('aria-selected', selected ? 'true' : 'false'); item.tabIndex = selected ? 0 : -1; });
    panels.forEach(function (panel) { panel.hidden = panel.id !== tab.getAttribute('aria-controls'); });
    tab.focus();
  }
  tabs.forEach(function (tab, index) {
    tab.addEventListener('click', function () { activateTab(tab); });
    tab.addEventListener('keydown', function (event) {
      var nextIndex = index;
      if (event.key === 'ArrowRight') nextIndex = (index + 1) % tabs.length;
      if (event.key === 'ArrowLeft') nextIndex = (index - 1 + tabs.length) % tabs.length;
      if (nextIndex !== index) { event.preventDefault(); activateTab(tabs[nextIndex]); }
    });
  });
