## Question 1: String `substring`

```java
String original = "DEFGHI";
String result = /* missing code */;
System.out.println(result);
```

Which expression makes this print `"HIDE"`?

- **A:** `original.substring(4) + original.substring(0, 1)`
- **B:** `original.substring(4) + original.substring(0, 2)`
- **C:** `original.substring(5) + original.substring(1, 2)`
- **D:** `original.substring(5) + original.substring(1, 3)`

**I guessed C. The correct answer is B.**

### Why B is correct

| Index | 0 | 1 | 2 | 3 | 4 | 5 |
|---|---|---|---|---|---|---|
| Char | D | E | F | G | H | I |

- `substring(4)` starts at index 4 and goes to the end, so it returns `"HI"`.
- `substring(0, 2)` returns indexes 0 to 1, so it returns `"DE"`.
- `"HI" + "DE"` gives `"HIDE"`.

### Why C was wrong

`substring(5)` is only `"I"`, and `substring(1, 2)` is only `"E"`. That gives `"IE"`. I didn't check the indexes.

**Takeaway:** `substring(a, b)` includes index `a` but stops before index `b`. I should write out the index table every time.

## Question 2: if / else if conditions

The game prints:

- `"Great!"` for scores greater than 500
- `"Good!"` for scores from 250 to 500, inclusive
- `"Try again."` for scores less than 250

**I guessed B. The correct answer is A.**

### Why A is correct

```java
String message = "";
if (score < 250)
{
   message = "Try again.";
}
else if (score <= 500)
{
   message = "Good!";
}
else
{
   message = "Great!";
}
System.out.println(message);
```

This is one connected chain, so only one branch runs.

- Below 250 prints "Try again."
- If that fails, the score is at least 250, so `score <= 500` means 250 to 500 and prints "Good!"
- Anything else is over 500 and prints "Great!"

### Why B was wrong

B uses a separate `if` for the second check. Its first `if` sets "Try again.", but the second `if (score <= 500)` runs no matter what. A score of 100 is also `<= 500`, so the message gets overwritten with "Good!". Scores under 250 never keep "Try again."

**Takeaway:** `if` followed by `else if` is one chain. Two separate `if` statements both get checked. I should test each choice with a low, middle, and high score such as 100, 250, 500, and 501.
